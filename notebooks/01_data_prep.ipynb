```python
# Imports
import os
import pandas as pd
import numpy as np
```

```python
# Loading & Inspection
red   = pd.read_csv('../data/raw/winequality-red.csv',   sep=';')
white = pd.read_csv('../data/raw/winequality-white.csv', sep=';')

# Quick inspection  — shape, nulls, duplicates & columns
for name, dfr in [('Red', red), ('White', white)]:
    dupes = dfr.duplicated().sum()
    print(f"{name} wine --> shape: {dfr.shape}  |  nulls: {dfr.isnull().sum().sum()}  |  duplicates: {dupes} ({dupes/len(dfr)*100:.1f}%) | columns: {list(dfr.columns)}")
```

```python
#Tagging wine types before Combining
red['wine_type'] = 'Red'
white['wine_type'] = 'White'

df = pd.concat([red, white], ignore_index=True)

print(f"Combined shape: {df.shape}")
print(df['wine_type'].value_counts())
```

```python
# Creating New Columns For Analysis

# Wine quality tier / ranking
def quality_tier(score):
    if score <= 5:
        return 'Low'
    elif score <= 7:
        return 'Medium'
    else:
        return 'High'

df['quality_tier'] = df['quality'].apply(quality_tier)

# QC pass/fail - binary column that indicates if it meets minimum industry standards
df['qc_status'] = df['quality'].apply(lambda x: 'Pass' if x >= 6 else 'Fail')

# SO2 binding ratio —> free SO2 / total SO2 | Used to measure preservation efficiency.
df['SO2_binding_ratio'] = df['free sulfur dioxide'] / df['total sulfur dioxide']

print(df['quality_tier'].value_counts())
print(df['qc_status'].value_counts())
print(f"\nSO2 binding ratio range: {df['SO2_binding_ratio'].min():.3f} — {df['SO2_binding_ratio'].max():.3f}")
```

```python
# Final check before saving
print(f"Final dataset shape: {df.shape}  | nulls: {df.isnull().sum().sum()} "
      f"\ncolumns: {list(df.columns)}"
      f"\n data types: {df.dtypes}"
      f"\n\n SUMMARY STATISTICS") 
df.describe().round(2)
```

```python
#Save to file
file = '../data/processed/wine_combined.csv'
df.to_csv(file, index=False)
print("✅ Saved to data/processed/wine_combined.csv")
```
